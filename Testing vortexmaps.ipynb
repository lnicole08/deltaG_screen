```python
import os
import pandas as pd
import vortexmap
```

```python
homecomp = "D:"
specifiedpath = homecomp

osardir = "\\ACC Lab Dropbox\\ACC Lab\\Nicole Lee\\Data Compilation\\osar_compiled\\Bootstrapped stats\\"
climbdir = "\\ACC Lab Dropbox\\ACC Lab\\Nicole Lee\\Data Compilation\\Climbing_New\\Compilation with delta\\2025deltagcollection\\"

osarpath = specifiedpath + osardir
climbpath = specifiedpath + climbdir
```

```python
sample = pd.read_csv(osarpath + "MB011B x Chrimson2_bootstrap.csv")
print(sample.columns.tolist())
print(sorted(sample["Metric"].unique()))
print(sorted(sample["Light_Intensity"].unique()))
```

```python
METRICS = sorted(sample["Metric"].unique())

osardata = vortexmap.load_bootstrap_data(osarpath, METRICS, responders=["Chrimson2"])
spirals, mean_vals, mbons = vortexmap.build_vortex_df(osardata, "Chrimson2", METRICS, None, chop_tail=2.5)
print(spirals.shape, mean_vals.shape, len(mbons))
```

```python
fig, ax = vortexmap.vortex_map(spirals, mean_vals)
ax.set_xticklabels(METRICS, rotation=45, ha="right")
ax.set_yticklabels(mbons, rotation=0)
```

```python
climbfiles = [f for f in os.listdir(climbpath) if f.endswith("_bootstrap.csv")]
print(len(climbfiles), climbfiles[:5])
```

```python
climbsample = pd.read_csv(climbpath + climbfiles[0])
print(climbsample.columns.tolist())
print(sorted(climbsample["Metric"].unique()))
```

```python
CLIMBMETRICS = sorted(climbsample["Metric"].unique())

climbdata = vortexmap.load_bootstrap_data(climbpath, CLIMBMETRICS)
responder = list(climbdata)[0]
spirals_c, mean_vals_c, mbons_c = vortexmap.build_vortex_df(climbdata, responder, CLIMBMETRICS, None, chop_tail=2.5)

fig, ax = vortexmap.vortex_map(spirals_c, mean_vals_c)
ax.set_xticklabels(CLIMBMETRICS, rotation=45, ha="right")
ax.set_yticklabels(mbons_c, rotation=0)
```
