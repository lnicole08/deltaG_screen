```python
df = pd.concat([dfe['Genotype'], dfe['Laid'], dfe['Eclosed']], axis = 1)

if df['Genotype'].str.contains("eOPN3").any():
   df['Genotype'] = df['Genotype'].str.replace("eOPN3", "OPN3")

# if df['Genotype'].str.contains("ACR").any():
#    df['Genotype'] = df['Genotype'].str.replace("ACR", "GtACR1")
   
# if df['Genotype'].str.contains("[ATR]").any():
#    df['Genotype'] = df['Genotype'].str.replace(" [ATR]", "*")

df['number'] = pd.factorize(df.Genotype)[0].tolist()

repeats = int(df[df.number == 1].number.value_counts().values[0])
```

```python
dftotal = pd.DataFrame()
dfnumbers = pd.DataFrame()
for n in df['Genotype'].unique():
    df1 = pd.DataFrame()
    df2 = pd.DataFrame()
    embryos = int(df[df['Genotype'] == n]['Laid'].sum())
    eclosed = int(df[df['Genotype'] == n]['Eclosed'].sum())
    
    #total embryos
    df1[n+" (H)"] = [1]*embryos
    df2[n+"_embryos"] = [embryos]
    
    #total pupate
    failpupate =embryos-eclosed                        
    df1[n + " (E)"] = [0]*failpupate + [1]*eclosed 
    df2[n + "_eclose"] = [eclosed]
    
    df1['ID'] = range(1, len(df1)+1)
    # dftotal = pd.concat([dftotal,df1], axis =1).reset_index(drop = True)
    dfnumbers = pd.concat([dfnumbers,df2], axis =1).reset_index(drop = True)
    
    dfmelted = pd.melt(df1, id_vars = ["ID"], value_vars=df1.columns[:-1])

    dftotal = pd.concat([dfmelted, dftotal], axis = 0).reset_index(drop = True)
    

dftotal['lethality'] = dftotal['value'] = 1 - dftotal['value']
```

```python
groups = {
    'A1ACR': (
        ('w1118 x A1ACR (H)', 'w1118 x A1ACR (E)'),
        ('elav x A1ACR (H)', 'elav x A1ACR (E)'),
        ('w1118 x A1ACR [ATR] (H)', 'w1118 x A1ACR [ATR] (E)'),
        ('elav x A1ACR [ATR] (H)', 'elav x A1ACR [ATR] (E)'),
    ),
    'Hf1ACR / Hf1ACR1': (
        ('w1118 x Hf1ACR (H)', 'w1118 x Hf1ACR (E)'),
        ('elav x Hf1ACR (H)', 'elav x Hf1ACR (E)'),
        ('w1118 x Hf1ACR1 [ATR] (H)', 'w1118 x Hf1ACR1 [ATR] (E)'),
        ('elav x Hf1ACR1 [ATR] (H)', 'elav x Hf1ACR1 [ATR] (E)'),
    ),
    'KALI1': (
        ('w1118 x KALI1 (H)', 'w1118 x KALI1 (E)'),
        ('elav x KALI1 (H)', 'elav x KALI1 (E)'),
        ('w1118 x KALI1 [ATR] (H)', 'w1118 x KALI1 [ATR] (E)'),
        ('elav x KALI1 [ATR] (H)', 'elav x KALI1 [ATR] (E)'),
    ),
    'KCR / KCR1': (
        ('w1118 x KCR (H)', 'w1118 x KCR (E)'),
        ('elav x KCR (H)', 'elav x KCR (E)'),
        ('w1118 x KCR [ATR] (H)', 'w1118 x KCR [ATR] (E)'),
        ('elav x KCR1 [ATR] (H)', 'elav x KCR1 [ATR] (E)'),
    ),
    'Chrimson': (
        ('w1118 x Chrimson (H)', 'w1118 x Chrimson (E)'),
        ('elav x Chrimson (H)', 'elav x Chrimson (E)'),
    ),
    'Unmatched': (
        ('w1118 x AsOPN3 [ATR] (H)', 'w1118 x AsOPN3 [ATR] (E)'),
        ('w1118 x elav (H)', 'w1118 x elav (E)'),
    ),
}
color_wt    = "#000000"
color_trans = "#00AEFF"
color_gray  = "#808080"

common_style = {'fontsize_rawxlabel': 9, 'fontsize_rawylabel': 9, 'fontsize_contrastxlabel': 9,
                'fontsize_contrastylabel': 9, 'fontsize_delta2label': 9, 'contrast_marker_size': 6}

n_groups = len(groups)
fig, axes = plt.subplots(n_groups, 1, figsize=(8, 6 * n_groups),
                         gridspec_kw={'hspace': 0.7})

dabest_objs  = {}
results_dict = {}

for ax, (name, idx) in zip(axes, groups.items()):
    obj = dabest.load(dftotal, idx=idx,
                      x="variable", y="lethality",
                      proportional=True, paired="baseline", id_col="ID")
    dabest_objs[name] = obj

    obj.mean_diff.plot(
        custom_palette={0: color_gray, 1: color_trans},
        contrast_ylim=(-0.3, 1.2),
        raw_label=f"Lethality proportion\n{name}",
        contrast_label="Δ Lethality proportion",
        ax=ax,
        **common_style,
    )

    results_dict[name] = obj.mean_diff.results

plt.show()
```
